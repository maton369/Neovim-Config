Verify that plot images do not overlap with Neo-tree sidebar.

Pass: image stays within editor pane, no overlap with Neo-tree.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# 1. Simple line plot (small)
x = np.linspace(0, 2 * np.pi, 100)
plt.figure(figsize=(5, 3))
plt.plot(x, np.sin(x), label='sin')
plt.plot(x, np.cos(x), label='cos')
plt.legend()
plt.title('sin / cos')
plt.tight_layout()
plt.show()
```

```python
# 2. Heatmap (wide — previously caused overlap)
data = np.random.randn(8, 12)
plt.figure(figsize=(10, 4))
plt.imshow(data, cmap='viridis', aspect='auto')
plt.colorbar()
plt.title('Heatmap (wide)')
plt.tight_layout()
plt.show()
```

```python
# 3. Subplots (large)
fig, axes = plt.subplots(2, 2, figsize=(8, 6))
for ax in axes.flat:
    ax.scatter(np.random.randn(30), np.random.randn(30), alpha=0.6)
    ax.set_title('scatter')
plt.suptitle('2x2 Subplots')
plt.tight_layout()
plt.show()
```

```python
# 4. Multiple images in one cell
# Each plt.show() sends a separate display_data to the kernel.
# Without plt.show() after each figure, only the last one appears.
for i, cmap in enumerate(['viridis', 'plasma', 'inferno']):
    plt.figure(figsize=(4, 3))
    plt.imshow(np.random.randn(6, 8), cmap=cmap, aspect='auto')
    plt.title(f'Plot {i+1}: {cmap}')
    plt.colorbar()
    plt.tight_layout()
    plt.show()  # required: omitting this shows only the last plot
```
